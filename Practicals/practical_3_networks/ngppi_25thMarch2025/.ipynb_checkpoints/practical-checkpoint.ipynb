Barabási, A-L. (2016). Network Science. Cambridge University Press. Retrieved from http://networksciencebook.com/

$G=(V,E)$ formally consists of a:
* set of vertices $V = \{i \in \{1,2,...,n\}\}$ and
* a set of "pairwise" edges $E = \{a_{ij} : i,j \in \{1,2,...,n\}\} \,\, or \,\, \{(i,j) : i,j \in \{1,2,...,n\}\}$ between them. An edge $a_{ij}$ connects vertex $i$ with vertex $j$.


Matrix representation $A = \{a_{ij}: i,j \in \{1,2,3....n\}\}$:
* $a_{ij} = 0$: No link
* if $a_{ij}\in \{0,1\}$: Unweighted
* if $a_{ij}\in \{0,c\} \,|\, c\in \mathbb{R}$: Weighted
* if $a_{ij}=a_{ji}$: Symmetric matrix or Undirected network
* if $a_{ij} \neq a_{ji}$: Asymmetric matrix or Directed network
* usually(but not always the case) self loops $a_{ii}$ are set to zero

Note: In networkX library, $a_{ij}$ refers to link from $i$ to $j$. But in some literature, books, papers, some authors use otherwise to denote the direction of link. We need to remember to keep a lookout for the details such as this.

## Average shortest path length, $L$
Given undirected graph and denoting $d_{ij}$ as the shortest path length(or geodesic distance) between nodes $i$ and $j$, the charactersitic path length or average path length or mean geodesic distance is defined as:
\begin{align*}
    L &= \frac{1}{n.(n-1)} \sum_{ij, \,i \neq j} d_{ij}\\
    &= \frac{2}{n.(n-1)} \sum_{ij, \, i < j} d_{ij}
\end{align*}

Note: $1/d_{ij}$ leads to concepts of efficiency.


```python
def compute_avg_short_path_length(G):
    vertices = list(G.nodes())
    num_nodes = G.number_of_nodes()
    
    path_length_sum = 0
    for i in range(num_nodes):
        for j in range(i+1, num_nodes):
            path_length_sum += nx.shortest_path_length(G, vertices[i], vertices[j])
    
    return 2*path_length_sum/(num_nodes*(num_nodes-1))
```

## Calculate diameter of the network, D

It is the longest among all shortest path lengths between all pairs of nodes.
$D = max\{d_{ij}: 0\leq i,j \leq n \, \wedge\, i<j\}$


```python
# diameter of a network as the longest of all the calculated shortest paths in a network.

# max(shortest path lengths from every node to every other node)

# can be computed only for connected graph

# manually written code
 
# understand the formula for diameter given in class slide
# and see whether the code below checks out or not    

def compute_diameter(G):
    max_dij_list = []
    nodes = list(G.nodes()) 
    
    for i in range(G.number_of_nodes()):
        nodeA = nodes[i]
        # self loops are removed
        # (i, j) and (j, i) are both same
        j_list = list(range(i+1, G.number_of_nodes()))
        dij_list = []
        
        for j in j_list:
            nodeB = nodes[j]
            dij_list.append( nx.shortest_path_length(G, nodeA, nodeB) )
        
        # check
        if len(dij_list)!=0:
            max_dij_list.append( max(dij_list) )    
    
    return max( max_dij_list ) 

```

## Clustering coefficient

* BASIC IDEA: The clustering coefficient is a measure of an "all-my-friends-know-each-other" property. This is sometimes described as the measure of how your friends(one hop neighbors) are also 'friends'(one hop neighbors) to each other. 

Local clustering coefficient $c_i$ of node $i$ refers to how likely $a_{jm} = 1$ for two neighbors $j$ and $m$ of node $i$. 

Given $G_i$ is the subgraph formed by the neighbors of node $i$, $e_i$ be the links between the neighbors of node $i$ denoted as $N_i = \{j: a_{ij}\in E \,\vee\, a_{ji} \in E \}$. It is computed as follows:

\begin{align*}
    c_i &= \frac{2\,.\,|e_i|}{k_i\,.\,(k_i-1)}\\
      &= \frac{2\,.\,|\{a_{jk}: j, k \in N_i, a_{jk}\in E\}|}{k_i\,.\,(k_i-1)}\\
      &=\frac{2\,.\,\sum_{j,m}a_{ij}\,a_{jm}\,a_{im}}{k_i\,.\,(k_i-1)}
\end{align*}

So in simpler words,  $$c_i = \frac{\text{Actual edges between neighbors of node $i$}}{\text{All possible edges between the neighbors of node $i$ }} $$ which has the same formulation in terms of network density. 



Thus, the average clustering coefficient of the entire graph is simply:
\begin{equation*}
    C = <c> = \frac{1}{n}\,\sum_{i=1}^n \, c_i
\end{equation*}


* The clustering coefficient for the entire network is the average of the clustering coefficients of all the nodes.
* A high clustering coefficient for a network is another indication of a small world. 



```python
# documentation
# https://networkx.org/documentation/stable/reference/algorithms/generated/networkx.algorithms.cluster.clustering.html#networkx.algorithms.cluster.clustering

# compute clustering coefficient of a specified node

print(nx.clustering(G1, nodes[10]))
```

```python
# manually compute clustering coefficient of a specific node

# Understand the formula for clustering coefficint of a node and compare with the code below

# credits: https://stackoverflow.com/questions/58044012/how-to-calculate-clustering-coefficient-of-each-node-in-the-graph-in-python-usin

def clust_coeff(network, node):
    neighbours=[n for n in nx.neighbors(network,node)]
    n_neighbors=len(neighbours) # same as degree of node
    n_links=0
    if n_neighbors>1:
        for node1 in neighbours:
            for node2 in neighbours:
                # has_edge checks if the graph has that particular edge or not
                                
                if network.has_edge(node1,node2):
                    n_links+=1
        n_links/=2 #because n_links is calculated twice
        clustering_coefficient=n_links/(0.5*n_neighbors*(n_neighbors-1))
        return clustering_coefficient
    else:
        return 0

clust_coeff(G1, 'Q92560')
```

# Assortativity

#### The hubs (nodes with very high degree) tend to link to each other and avoid linking to small-degree nodes while small-degree nodes tend to connect to other small-degree nodes.

But it is the opposite in the case of disassortative networks while in neutral networks the wiring is completely random. 
K nearest neighbours or Average degree connectivity of graph($k_{nn}(k)$) is simply the average nearest neighbours with degree k. In mathematical terms:


\begin{equation}\label{knn1}
    k_{nn}(k) = \sum_{k'} k' P(k'|k)
\end{equation}


where $P(k'|k)$ is the conditional probability starting from a link of k degree reaching a node of k' degree. From the log-log plot of $k_{nn}(k)$ as a function of k, the equation 


\begin{equation}\label{knn2}
    k_{nn}(k) = Ck^\mu
\end{equation}


is to be fit so as to obtain the parameter $\mu$ known as **correlation exponent**. The sign of this exponent decides the type of assortativity observed in our network. 

- If $\mu >0$ then it is an **Assortative network**. 
- When $\mu=0$ representing that $k_{nn}$ is independent of k, it can be conclude that this is a **Neutral network**. 
- Lastly, if $\mu<0$, this corresponds to a **Disassortative network**.  

For More information: http://networksciencebook.com/chapter/7#assortativity

## Examples







```python
# Assortativity plot code
from scipy.optimize import curve_fit
from sklearn.metrics import r2_score


def logKnn(logK, mu, c):
    return mu*logK + c


def plotnfit_Knn(G, degree_direction_source,degree_direction_target):
    """Plots a log-log scatter plot of Knn(k) vs k and also does fitting. 
    Parameters
    ----------
    G : Networkx graph
       A graph
    degree_direction: str, 'in'-> in-degree K_in, 'out'-> out-degree K_out
    """
    #knn = nx.k_nearest_neighbors(G, source=degree_direction_source,
    #                             target=degree_direction_target,weight='weight')
    knn = nx.average_degree_connectivity(G, source=degree_direction_source,
                                 target=degree_direction_target,weight='weight')
    
    # Extract Knn(K) and K
    k = np.array(list(knn.keys()))
    #print(max(k))
    knn_k = np.array(list(knn.values()))

    # remove zeros from Knn, clean the data since we are applying log function.
    knn_new = knn_k[knn_k>0]
    k_new = k[knn_k>0]

    # apply log to do log fitting instead of power law fitting
    x = np.log(k_new)
    y = np.log(knn_new)
    params,cov = curve_fit(logKnn,x,y ) # cov is the covariance matrix
    mu = params[0]
    c = params[1]

    # compute r2, fit to existing data 
    y_pred = logKnn(x,*params)
    print( r"$(\mu, c, R^2$): (%.3f, %.3f, %.3f)"%(mu, np.exp(c), r2_score(y, y_pred)) ) 

    # fit to new data
    x = np.log(np.linspace(min(k_new), max(k_new), 100))
    yfit = logKnn(x,*params)

    plt.figure(figsize=(4,3))
    
    plt.plot(k,knn_k,'.')
    plt.axhline(y=np.mean(knn_k),c='black',lw=0.95,linestyle='-',label='neutral behaviour')
    plt.plot(np.exp(x),np.exp(yfit),'--',lw=2.3,dashes=[3,6],label='$\\sim k^{%.3f}$'%(mu))#,degree_direction_source))
    
    
    plt.yscale('log')
    plt.xscale('log')
    
    # aesthetics
    plt.tick_params(direction='in',bottom=True,top=True,left=True,right=True)
    plt.grid(linestyle='--',alpha=0.5,which='both')
    plt.legend(loc='best',prop={'size': 13})
    
    return mu, np.exp(c), r2_score(y, y_pred)
    
```

# Degree distribution, Power law, Scale free networks

The degree distribution when said to follow power law, can be quantified by the power law exponent $\gamma$. 

\begin{equation}
 p_k \sim k^{-\gamma}
\end{equation}

In the log-log scale, the data points(degrees) set to form a straight line

\begin{equation}
 log (p_k) \sim {-\gamma}log(k)
\end{equation}

**Meaning of Scale free** 

n-th order moment of degree distribution is given by:

\begin{align}
    <k^n> = \sum_{k_{min}}^{k_{max}}k^n\,p_k\, \approx \int_{k_{min}}^{k_{max}}k^n\,p(k)\,dk\\
    =C\frac{k_{max}^{n-\gamma+1}-k_{min}^{n-\gamma+1}}{n-\gamma+1}
\end{align}

 - $\gamma = 2$, $<k>$ is finite, while $<k^n>$ such that n=2,3,4....diverges. i.e. standard deviation diverges.
 - $\gamma = 3$, $<k>, <k^2>$  is finite. while higher order moment diverges.
 - In the range $2<\gamma<3$, we have node degrees $<k> \pm\, stdev. = <k>\pm\, \infty $ that can be very high or very low. The degree distribution is heterogeneous.

This means that when we randomly choose a node, we do not know what to expect: The selected node’s degree could be tiny or arbitrarily large. Hence networks with $\gamma < 3$ do not have a meaningful internal scale, but are "scale-free". 

Counter example is Random network(recall Erdos Renyi network from earlier). 

How to say when a network follows Power law?
- Many nodes with few links
- a few hubs with large number of links


**How do we know if a network is scale-free?**
On one end, a quick look at the degree distribution will immediately reveal whether the network could be scale-free: In scale-free networks the degrees of the smallest and the largest nodes are widely different, often spanning several orders of magnitude. In contrast, these nodes have comparable degrees in a random network. As the value of the degree exponent plays an important role in predicting various network properties, we need tools to fit the pk distribution and to estimate γ. 

**A scale-free network is a network whose degree distribution follows a power law and $\gamma$ parameter is in the range $2 < \gamma < 3$.** 




```python
# plotting degree distribution

# plot degree distribution with constant automated binning

def plot_constbin(degree):
    degree = np.array(degree) # if list convert to np array
    deg = degree[degree>0] # don't consider values=0

    edges= np.histogram_bin_edges(deg,bins='auto')
    pk = np.array(plt.hist(deg,bins=edges,density=True)[0])

    plt.close()

    edges2 = list(edges[1:])+[0]

    mid_point = (np.array([a+b for a,b in zip(edges,edges2)])/2)[:-1]
    plt.figure(figsize=(5,3))
    
    plt.loglog(mid_point, pk,'.',color="orangered")
    
    plt.xlabel("x",size=15)
    plt.ylabel("$P(x)$",size=15)
    plt.title("log-log plot with automated \n optimized x-axis binning, bin size = %.2f"%(edges[1]-edges[0]),
              size=14)
    #print("bin size: ",(edges[1]-edges[0]))
    
    plt.tick_params(direction='in',bottom=True,top=True,left=True,right=True)
    plt.grid(linestyle='--',alpha=0.5,which='both')
```

```python
# Plot and fit, find gamma
import powerlaw

from statsmodels.distributions.empirical_distribution import ECDF
from scipy import stats
from sklearn.metrics import r2_score


def ppareto(x, shape, location):
    # x is numpy array
    return (x >= location) * (1 - (location / x)**shape)


def plotnFit_ccdf(degree, fit=True, discrete=True):
    degree = np.array(degree) # if list convert to np array
    degree = degree[degree>0] # don't consider values=0
    ecdf = ECDF(degree)
    
    plt.figure(figsize=(5,3))

    
    plt.loglog(degree, (1-ecdf(degree)),'.', label="Observed data",alpha=0.4)
    
    if fit==True:
            results = powerlaw.Fit(degree, discrete=discrete, estimate_discrete=discrete)
            alpha = results.power_law.alpha
            xmin =  int(results.power_law.xmin)
            sigma = results.power_law.sigma
            print("gamma: ",alpha)
            print("sigma: ",sigma)# standard error
            print("kmin: ", xmin)
            
            shape = alpha-1
            location = xmin

            data = np.array(list(np.arange(min(degree), max(degree))))
            x = data[data>=location]

            # multiply ccdf_model with (1-ecdf(location))
            ccdf_model = (1-ecdf(location))*(1-ppareto(x, shape, location)) # ppareto is CDF

            y = 1-ecdf(degree[degree>=location])
            y_pred = (1-ecdf(location))*(1-ppareto(degree[degree>=location], shape, location))
            r2 = r2_score(y, y_pred)    
            print("R2:", r2_score(y, y_pred))

            plt.loglog(x, ccdf_model, ls='--', c="red",linewidth=4, 
                    label="$\\alpha = %.3f \\pm %.3f$"%(alpha,sigma) )

            plt.axvline(x=location, ls="--", linewidth=1, c="green", label="$k_{min}$ = %d"%location)

    
    plt.tick_params(direction='in',bottom=True,top=True,left=True,right=True)
    plt.grid(linestyle='--',alpha=0.5,which='both')
    plt.legend(loc="best",prop={'size': 13})
    
    plt.xlabel("x",size=15)
    plt.ylabel("CCDF, P[X > x]",size=15)
    #return alpha, sigma, xmin, r2
    
    #plt.title("Weighted PPI-LCC",size=15)
```

```python
# list of degrees of nodes
degree_list = list(dict( G1.degree() ).values())

plotnFit_ccdf(degree_list)
plt.savefig(image_path+"fitdistr_ngppi.pdf", \
        facecolor="white", bbox_inches="tight", dpi=600)
```

### Conclusions: 

* Doesn't follow power law for the entire network. 

* Only from k_min > 97, it truly follows power law. 
* Network is also not "Scale free"

* However, the degree distribution is still heterogeneous i.e. many nodes with fewer links but few nodes with very high links.

# Community detection

Communities are a property of many networks in which a particular network may have multiple communities such that nodes inside a community are densely connected. Nodes in multiple communities can overlap. 

Example: Facebook, instagram - Closely-knit interaction with your friends, colleagues, family members and a few other important people in your life. They form a very dense community inside your social network.

#### Main message: Network nodes are tightly connected in knit groups within communities and loosely connected between communities.

Community detection depends on single attribute type which are **edges**. 

There are various community detection algorithms. Details of it will be taught in Network Science Course from DEI, UNIPD. 

```python
# A simple example to understand

G_simple = nx.Graph()
#G_simple.add_nodes_from(range(10))

l1 = [0,1,2,3,4]
for i in range(len(l1)):
    j_list =  range(i+1, len(l1))
    for j in j_list:
        G_simple.add_edge(l1[i], l1[j])

l2 = [5, 6, 7, 8, 9]
for i in range(len(l2)):
    j_list =  range(i+1, len(l2))
    for j in j_list:
        G_simple.add_edge(l2[i], l2[j])

G_simple.add_edges_from([(1,6),(3,8)])

nx.draw_networkx(G_simple, pos=nx.spring_layout(G_simple)) 
```

# Visualization of few organelles

```python
#np_complex_nodes = pd.read_csv("Nuclear pore complex nodes.csv")['nodes'].to_list()
para_nodes = pd.read_csv("Paraspeckle nodes.csv")['nodes'].to_list()
nucl_speckle_nodes = pd.read_csv("Nuclear speckle nodes.csv")['nodes'].to_list()
```

```python
# Select nodes from organelles
comm_nodes = para_nodes  + nucl_speckle_nodes

# remove all the nodes except from these organelles
delete_nodes = set(list(G1.nodes()))-set(comm_nodes)

# Note: remember to use nx.Graph() on it since it doesn't work otherwise.
G_small = nx.Graph( deepcopy(G1) )

G_small.remove_nodes_from(  list(delete_nodes) )

edge_listA, edge_listB = zip(*list(G_small.edges()))
```

```python
# Storing the data

community = []

for node in list(G_small.nodes()):
    if node in set(nucl_speckle_nodes).intersection(set(para_nodes)) :
        community.append(0)

    elif node in nucl_speckle_nodes:
        community.append(1)
    
    elif node in para_nodes:
        community.append(2)
    
    
```

```python
df_Comm = pd.DataFrame({"Source":edge_listA, "Target":edge_listB})
df_Comm.to_csv("organelle_edges_visualization.csv", index=False)

df_Comm_node = pd.DataFrame({"node": list(G_small.nodes()) , 
                             "community": community }, )

df_Comm_node.to_csv("organelle_nodes_visualization.csv", index=False)
```
